```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy.modeling import models, fitting
from get_MJD import fit_gauss, multi_gauss
from matplotlib import rcParams
import os
import matplotlib.ticker as ticker
```

```python
def burst_mjds(database, lband=True):

    #Load in the dataframe with all the stats
    stats_df = pd.read_csv(database, sep=",", header=0, index_col=0, na_values='NA', engine='python')
    
    #Grab all unique burst ids
    ids = stats_df["id"].unique()
    
    #bursts that are duplicates -- filter these, otherwise we count them double
    ex_burst = ["B68-st", "B69-st", "B85-st", "B107-st", 'B130-tr']
    stats_df = stats_df[~stats_df['id'].isin(ex_burst)]
    
    if lband == True:
        stats_df = stats_df[(stats_df['f0_MHz'] != 328.0)]
    
    else:
        stats_df = stats_df[(stats_df['f0_MHz'] == 328.0)]
        
    #display(stats_df)
    #Create a list with the telescope names for easy filtering later on
    exps_tel = [i.split('-')[1] for i in ids]
        
    #Waits is in seconds
    multicomp = []
    waits = []
        
    for burst_id, telescope in zip(ids, exps_tel):
        
        #Only take the SFXC data, since we use that for the arrival times
        if telescope != 'st':

            df_burst_id_temp = stats_df[(stats_df['id'] == burst_id) & (stats_df['src'] == 'sfxc')]

        #In the case of St; just use default lines
        elif telescope == 'st':

            df_burst_id_temp = stats_df[stats_df['id'] == burst_id]
        
        #Sort on component, since the zero-comp arrives first; otherwise you get negative results
        df_burst_id_temp = df_burst_id_temp.sort_values(by=['component'])
        
        #Component seperation is calculated based on the column: toa_mjd_utc which is in MJD
        num_comp = len(df_burst_id_temp)            

        if num_comp > 1:
            
            #Retrieving the MJD of the components
            times = df_burst_id_temp['toa_mjd_utc'].values
            
            #Getting the wait times, converting to seconds
            for wait in np.diff(times):

#                 if 0.0013 < wait*86400 < 0.0016:
#                     print(burst_id,wait*86400)
#                     #display(df_burst_id_temp)
                waits.append(wait*86400)
                
            multicomp.append(burst_id)

    return multicomp, waits
```

```python
multicomp_l, waits_l = burst_mjds(database="../dbs/burst_stats_r117.csv", lband=True)
print(len(waits_l))
```

```python
multicomp_p, waits_p = burst_mjds(database="../dbs/burst_stats_r117.csv", lband=False)
print(len(waits_p))
print(multicomp_p)
```

```python
def wait_plotter(waits):

    n, bins, _ = plt.hist(waits, 50)
    waits = np.log10(waits)
    
    # the below would create a log-log plot, but we only want log on the y-axis
    # and so we just get the bins right first
    n, bins, _ = plt.hist(waits, bins='sqrt', log=True)
    plt.clf()

    #plt.figure(figsize=(10, 8))
    f, ax = plt.subplots(1, figsize=(12,10))
    n, bins, _ = plt.hist(waits, bins, color='#f1a340', edgecolor = 'black', alpha=0.8)

    # fit gaussian
    amp = np.max(n)
    sigma = bins.std()
    x0 = np.median(bins)
    params = np.array([[amp, x0, sigma]])
    model = multi_gauss(params)

    diffs = np.diff(bins) / 2
    mbins = bins[:-1] + diffs
    fit, fit_lm = fit_gauss(model, n, mbins)

    print(fit)
    plotbins = np.arange(bins.min(), bins.max(), 0.01)
    plt.plot(plotbins, fit(plotbins), color="#998ec3", linewidth=5)
    plt.xlim(-1.9,-3.4)
    plt.ylim(0,27)
    plt.xlabel(r'$\mathrm{\log_{10}\left(Component\;separation / [second]\right)}$', fontsize=25)
    plt.ylabel('Count', fontsize=25)
    
    plt.xticks(fontsize=22)
    plt.yticks(fontsize=22)
    
    for axis in ['top','bottom','left','right']:
        ax.spines[axis].set_linewidth(2)
            
        ax.tick_params(axis='y', which='major', direction='out', length=10, width=2)
        ax.tick_params(axis='y', which='minor', direction='out', length=5, width=2)
        ax.tick_params(axis='x', which='major', direction='out', length=10, width=2)
        ax.tick_params(axis='x', which='minor', direction='out', length=5, width=2)
        
    
    ax.yaxis.set_major_locator(ticker.MultipleLocator(5))
    ax.yaxis.set_minor_locator(ticker.MultipleLocator(1))
    
    ax.xaxis.set_major_locator(ticker.MultipleLocator(0.2))
    ax.xaxis.set_minor_locator(ticker.MultipleLocator(0.1))
    
    plt.savefig('../plots/wait_time_hist_log.png', dpi=300)

    return
```

```python
wait_plotter(waits_l)
```
