Script to calculate the maximum energetics in terms of erg/s for our brightest burst component. (B77-st, component 2
)

The formula used is: 
E = (S[Jy] * 1e-23 * v[Hz] * 4pi* D_L^2[cm^2]) / (1+z)^(1+alpha) </br>
Units are [erg/s]

Formula from Pragya Chawla and later independently send by Stuart Ryder (via Clancy James) 

```python
import numpy as np
from astropy import units as u
```

```python
def isotropic_energies(flux, distance=362.4, redshift=0.0771, alpha=0):
    "Convert Jy to erg/s"

    flux_arr = np.array(flux) 

    dis = 1.0 * u.megaparsec
    megap_cm = dis.to(u.cm).value

    #mpc to cm
    distance_lum_cm = megap_cm*distance

    #redshift correction, see evernote for details
    #redshift is from: Ravi et al, DSA-110
    #distance is also from Ravi et al, FAST converts themselves based on PLANCK 2016
    #link: https://ui.adsabs.harvard.edu/abs/2023ApJ...949L...3R/abstract
    z = redshift
    red_cor = 1 / (1 + z)**(1 + alpha)

    #Assume canocical bandwidth of 1 GHz
    ghz = 1e9
    
    energy_iso = flux_arr * 4*np.pi*(distance_lum_cm**2) * 1e-23 * red_cor * ghz
    print(f"{energy_iso}")

    return 
```

In order to confirm that this method of calculating the flux density is correct, I confirmed the numbers by recalculating the result from Ryder et al (2023) [z=1 burst]

- Flux is 110 Jy (45 Jy ms / 0.41 ms) </br>
- Where 0.41 is the intrinsic width and not the observed width (!) </br>

- Redshift and distance are taken from the paper and since this is in the frame of the galaxy I presume you can assume alpha=0. 

```python
isotropic_energies(flux=110, distance=6929, redshift=1.016)
```
