```python
import matplotlib
matplotlib.rcParams['mathtext.fontset'] = 'custom'
matplotlib.rcParams['mathtext.fontset'] = 'stix'
matplotlib.rcParams['font.family'] = 'STIXGeneral'
```

```python
def func_powerlaw(x, N=1, gamma=1):
    return x**gamma * N
```

```python
def func_schechter_emcee(x, pars):
    '''Power-law function with an exponential cut-off.
       Inputs:
           x - input x value(s) (can be list or single value).
           pars - parameters, list of PL normalisation (at x = 1) and power-law index.'''
    (N,gamma,E_cut) = pars
    return N *((x/E_cut)**gamma)*np.exp(-x/E_cut)
```

```python
def plotting():

    # x_space = np.arange(0,100)
    # y = func_powerlaw(x_space, scale=1, gamma=1)
    # y_s = func_schechter_emcee(x_space)

    N = 13
    N_p = 4800
    gamma = -2
    ecut = 20
    
    x_range = np.linspace(1, 200, 100)
    y_s = func_schechter_emcee(x_range, [N, gamma, ecut])
    y_p = func_powerlaw(x_range, N_p, gamma)
    
    f, ax = plt.subplots(1, figsize=(12,10))

    #plt.plot(x_range, y_s, lw=10, c="#f1a340")
    
    plt.plot(x_range, y_p, lw=10, c="#998ec3")

    #plt.xlim(1,110)
    plt.xscale('log')
    plt.yscale('log')
    plt.xticks([])
    plt.yticks([])
    plt.xlabel("Energy", fontsize=55)
    for axis in ['top','bottom','left','right']:
        ax.spines[axis].set_linewidth(4)
        ax.tick_params(axis='y', which='major', direction='out', length=0)
        ax.tick_params(axis='y', which='minor', direction='out', length=0)
        ax.tick_params(axis='x', which='major', direction='out', length=0)
        ax.tick_params(axis='x', which='minor', direction='out', length=0)

    save_name="empty_powerlaw"
    plt.savefig(f"{save_name}.pdf", format = 'pdf', dpi=600, bbox_inches='tight')
    plt.savefig(f"{save_name}.png", \
            bbox_inches='tight', facecolor='white', transparent=False, dpi=300)
    
    plt.show()

plotting()
```
