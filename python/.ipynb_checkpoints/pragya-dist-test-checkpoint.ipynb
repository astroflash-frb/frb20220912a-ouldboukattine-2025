```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.stats import chi2

def get_poisson_error(r, conf_level):
    low_lim = 0.5 * chi2.ppf((1 - conf_level) / 2, 2 * r)
    low_lim[np.isnan(low_lim)] = 0
    upp_lim = 0.5 * chi2.ppf((1 + conf_level) / 2, 2 * (r+1))
    upp_lim[np.isnan(upp_lim)] = 0
    return low_lim, upp_lim

def get_rate_errors(bursts, conf_level=0.90):
    low_lim, upp_lim = get_poisson_error(bursts, conf_level=conf_level)
    # if no bursts are detected, get single sided upper limit as rate 
    lim = bursts == 0

    new_conf_level = 1 - ( (1-conf_level)*2 ) # for single sided upper limit 
    _, upp_lim[lim] = get_poisson_error(np.array([0]), conf_level=new_conf_level)
    
    dx_upp = upp_lim - bursts
    dx_low = bursts - low_lim

    return dx_low, dx_upp 

def histogram(
    arr,
    *args,
    func=None,
    distnm="test",
    bin_num=20,
    logx=False,
    logy=False
):
    """Plots histogram of given simulated property and also plots the theoretical distribution
       based on given function and arguments
       distnm: To be used as label for legend"""
    plt.figure()
    if logx == True:
        bins = np.logspace(np.log10(np.min(arr)), np.log10(np.max(arr)), bin_num)
        plt.gca().set_xscale("log")
    else:
        bins = np.linspace(np.min(arr), np.max(arr), bin_num)
    if logy == True:
        plt.gca().set_yscale("log")
    counts_norm, bins, _ = plt.hist(arr, bins=bins, density=True, label="Data") ###
    counts, _ = np.histogram(arr, bins=bins, density=False) ###
    
#     counts = counts.astype(np.float64)
#     counts_norm = counts_norm.astype(np.float64)
#     counts[counts == 0] = np.nan 
#     counts_norm[counts_norm == 0] = np.nan 

    norm_fac_counts = sum(counts) * np.diff(bins)
    #norm_fac_counts = counts/counts_norm ###
    # get errors
    error_low, error_upp = get_rate_errors(counts, conf_level = 0.68) ###
    # need to divide poissonian errors by normalisation that histogram function does
    error_low /= norm_fac_counts ###
    error_upp /= norm_fac_counts ###
    bin_centers = np.sqrt(bins[0:-1]*bins[1:]) ###
    plt.plot([bin_centers, bin_centers], [counts_norm - error_low, 
                                          counts_norm + error_upp], '-', color='C2') ###
    if func:
        # calculate total area under the curve 
        norm = quad(func, np.min(arr), np.max(arr), args=(args))[0]
        # get the value of the power law function for each bin 
        plt.plot(bins, func(bins, *args) / norm, label="%s" % distnm)
        # in order to normalise, divide function by total area under the curve 
        plt.legend(loc="best", fontsize=13)
        
def pl_dist_pdf(x, x_min, ind):
    """PDF of power law distribution"""
    return (ind - 1) * (x/x_min)**-ind / x_min

def pl_inv_cdf(u, x_min, ind):
    return x_min * ((1-u)**(1/(1-ind)))

def inv_trans_samp(N, func, *args):
    """Generate N samples based on the given inverse CDF (func) and corresponding arguments"""
    u = np.random.rand(N)
    x = func(u, *args)
    return x

N = 100
low_lim = 1e30 
index = 1.5
# sampling function to generate events distributed as a power law 
energetics = inv_trans_samp(N, pl_inv_cdf, low_lim, index+1)
histogram(energetics, low_lim, index+1, func=pl_dist_pdf, distnm = "Power Law", 
          bin_num = 20, logx = True, logy=True)
```
