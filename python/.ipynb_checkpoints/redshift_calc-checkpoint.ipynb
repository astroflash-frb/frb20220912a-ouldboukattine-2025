```python
import numpy as np
import astropy.units as u
import matplotlib.pyplot as plt
from astropy.cosmology import Planck15 as cosmo
```

```python
def energy(fluence_jy_ms, bw_ghz, z, alpha):
    
    fluence_jy_s = fluence_jy_ms * 1e-23 * 1e-3
    bw_h = bw_ghz * 1e9    
    energy = (fluence_jy_s * bw_h * 4 * np.pi * (cosmo.luminosity_distance(z).to(u.cm).value)**2) / (1+z)**(2+alpha)
    #print(energy)
    return energy

def ratio(energy, fluence_jy_ms, bw_ghz, alpha, plot=False):
    
    fluence_jy_s = fluence_jy_ms * 1e-23 * 1e-3
    bw_h = bw_ghz * 1e9

    r_side = energy / (4 * np.pi * bw_h * fluence_jy_s) # equal to d_l^2 / (1+z)^(2+alpha)
    #print('r_side',r_side)
    z_max=17
    z_range = np.linspace(0, z_max, 1500)
    lhs = cosmo.luminosity_distance(z_range).to(u.cm).value**2 / ( (1+z_range)**(2+alpha) )
    
    z_match = z_range[np.argmin(np.abs(lhs - r_side))]
    #print(z_match)
    
#     plt.figure(figsize=(10,6))
#     plt.plot(z_range, np.sqrt(lhs), color="#2166ac")
#     plt.axhline(np.sqrt(r_side), color='#b2182b')
#     plt.axvline(z_match, color='black', ls='-.')
#     plt.tick_params(labelsize=20)
#     plt.xlabel("Redshift (z)", fontsize=20)
#     plt.ylabel(r"$\sqrt{\dfrac{\mathrm{D_{L}}^2}{(1+\mathrm{z})^{2-1.5}}}$", fontsize=20, rotation=0,labelpad=85)
#     plt.xlim(-0.3, z_max)
#     #plt.ylim(0,1.5*np.sqrt(r_side))
#     plt.yscale('log')
#     #plt.legend(fontsize=20)
#     plt.show()
    
    if plot:
        plt.figure(figsize=(10,6))
        plt.plot(z_range, lhs, color="#2166ac")
        plt.axhline(r_side, color='#b2182b')
        plt.axvline(z_match, color='black', ls='-.')
        plt.tick_params(labelsize=20)
        plt.xlabel("Redshift (z)", fontsize=20)
        plt.ylabel(r"$\dfrac{\mathrm{D_{L}(z)}^2}{(1+\mathrm{z})^{2-1.5}}$", fontsize=20, rotation=0,labelpad=85)
        plt.xlim(-0.3, z_max)
        #plt.ylim(0,1.5*np.sqrt(r_side))
        plt.yscale('log')
        #plt.legend(fontsize=20)
        #plt.savefig("redshift_distance.png", bbox_inches='tight', facecolor='white', transparent=False, dpi=600)
        plt.show()
    
    return z_match
```

```python
#####
# L-band 
#####
energy_wb_l = energy(fluence_jy_ms=1587, bw_ghz=0.01, z=0.0771, alpha=0)
redshift_wb_l = ratio(energy=energy_wb_l, fluence_jy_ms=0.053, bw_ghz=0.01, alpha=-1.5)
print(f"emitting frequency: {(redshift_wb_l+1)*1.271}")
print(f"redshift: {redshift_wb_l}")

energy_wb_l = energy(fluence_jy_ms=1587*1.2, bw_ghz=0.01, z=0.0771, alpha=0)
redshift_wb_l_err_1 = ratio(energy=energy_wb_l, fluence_jy_ms=0.053, bw_ghz=0.01, alpha=-1.5)
print(f"plus error: {redshift_wb_l_err_1 - redshift_wb_l}")

energy_wb_l = energy(fluence_jy_ms=1587*0.8, bw_ghz=0.01, z=0.0771, alpha=0)
redshift_wb_l_err_2 = ratio(energy=energy_wb_l, fluence_jy_ms=0.053, bw_ghz=0.01, alpha=-1.5)
print(f"minus error: {redshift_wb_l_err_2 - redshift_wb_l}")
```

```python
#####
# P-band
#####

energy_wb_p = energy(fluence_jy_ms=2629, bw_ghz=0.054, z=0.0771, alpha=0)
redshift_p = ratio(energy=energy_wb_p, fluence_jy_ms=1, bw_ghz=0.054, alpha=-1.5)
print(f"emitting frequency: {(redshift_p+1)*0.328}")
print(f"redshift: {redshift_p}")

energy_wb_p = energy(fluence_jy_ms=2629*1.2, bw_ghz=0.054, z=0.0771, alpha=0)
redshift_p_err_1 = ratio(energy=energy_wb_p, fluence_jy_ms=1, bw_ghz=0.054, alpha=-1.5)
print(f"plus error: {redshift_p_err_1 - redshift_p}")

energy_wb_p = energy(fluence_jy_ms=2629*0.8, bw_ghz=0.054, z=0.0771, alpha=0)
redshift_p_err_2 = ratio(energy=energy_wb_p, fluence_jy_ms=1, bw_ghz=0.054, alpha=-1.5)
print(f"minus error: {redshift_p_err_2 - redshift_p}")
```
