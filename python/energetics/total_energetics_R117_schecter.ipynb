Let us assume the spectral burst energy density distribution takes the form of the Schecter function:

$N(E) = C \big(\frac{E}{E_{\rm char}}\big)^{-\gamma_{\rm D}} \exp\big(- \frac{E}{E_{\rm char}}\big)$

Limited by some $E_{\rm min}$ and optionally by some $E_{\rm max}$. Note that due to the exponential cut-off, as long as $E_{\rm max}$ > $E_{\rm char}$, the total energy does not depend strongly on the choice of $E_{\rm max}$. Then rearranging and using a constant we find:

We use an $E_{\rm min} = 3.3 \times 10^{30} \, {\rm erg \, Hz^{-1}}$, $E_{\rm char} = 2.09 \times 10^{32} \, {\rm erg \, Hz^{-1}}$, $\gamma_{\rm D} = -1.12$, and $C = 13.55$ from the paper. 


Then we can integrate over the function to find the total spectral energy in bursts.


From here we just multiply by the typical bandwidth to find the total energy in radio, then multiply by some efficiency to find the total overall energy, then divide by the number of unique observing hours to derive a total averaged luminosity. This is then compared to the approximate magnetospheric energy $U_{\rm B} \approx \frac{B^2 R_{\rm NS}^3}{8 \pi}$ to derive a deletion timescale

```python
import numpy as np
import scipy
import scipy.integrate as integrate

# minimum burst E/Hz
min_E = 3.3e30

# maximum burst E/Hz
max_E = 1e33

# characteristic energy E/Hz
E_char_true = 2.09e32

# powerlaw slope 
gamma_true = -1.12

# Normalisation factor
N_true = 13.55

# An assumed delta_nu (burst bandwidth)
dnu = 1e8

# An assumed radio efficiency
eff = 1e5

# Total number of unique hours on source
hours_on_source = 1769.24

# magnetospheric energy
mag_energy = 1e18*1e30/(8*np.pi)

def schecter(E, E_char, C, gamma):
    
    return C*(E/E_char)**(gamma) * np.exp(-E/E_char)


Total_spectral_energy_density = integrate.quad(schecter, min_E, max_E, args=(E_char_true, N_true,gamma_true))

Total_power = Total_spectral_energy_density[0]*dnu*eff/(hours_on_source)

print(Total_spectral_energy_density[0]*dnu)
print(Total_spectral_energy_density[0]*dnu/(hours_on_source))

print(Total_power, ' erg\Hr')

mag_energy = 1e18*1e30/(8*np.pi)

print('Runs out in: ', mag_energy/Total_power, ' hours')

```
