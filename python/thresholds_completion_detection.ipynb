### Simple script to calculate the different thresholds both completion and detection.

```python
import numpy as np
```

```python
def fluence_limit(sefd=420, width=1, bw=100, snr=20):
    '''
    sefd in Jy, width in ms, bw in MHz
    '''
    fluence_limit = snr * sefd * np.sqrt( (width/1000.) / (2 * bw * 1e6))
    fluence_limit_jyms = fluence_limit * 1000
    print(f"{fluence_limit_jyms:.1f} Jy ms")
    # return things in Jy~ms
    return 
```

# Detection Thresholds

    - Width of 1 ms, a typical FRB
    - bandwidth takes account of RFI and band edges
    - S/N ratio set at 7 which is the limit we use for Heimdall / Single_pulse_search

```python
#Westerbork P-band
fluence_limit(sefd=2100, width=1, bw=50, snr=7)
```

```python
#Westerbork L-band
fluence_limit(sefd=420, width=1, bw=100, snr=7)
```
